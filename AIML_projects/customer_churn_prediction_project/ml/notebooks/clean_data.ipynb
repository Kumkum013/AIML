```python
import pandas as pd

df = pd.read_excel('E:/AIML-main/sem4 ml project/ml/datasets/Telco_customer_churn.xlsx')
print("Raw shape:", df.shape)
```

```python
# drop irrelevant and leaky columns
df = df.drop(columns=[
    'CustomerID', 'Count', 'Country', 'State', 'City', 'Zip Code', 
    'Lat Long', 'Latitude', 'Longitude', 
    'Churn Label', 'Churn Score', 'Churn Reason'
])
print("Shape after dropping columns:", df.shape)
```

```python
# fix total charges the column causing trouble
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce')
print("Nulls in Total Charges before fix:", df['Total Charges'].isnull().sum())

df['Total Charges'] = df['Total Charges'].fillna(0)
print("Nulls in Total Charges after fix:", df['Total Charges'].isnull().sum())
```

```python
df.dropna()              # Drop rows with *any* missing values
df.dropna(axis=1)        # Drop columns with missing values
```

```python
df.duplicated() 
```

```python
df.drop_duplicates() 
```

```python
df.describe()
```

```python
df.dtypes
```

```python
df = df.rename(columns={'Churn Value': 'Churn'})
print(df['Churn'].value_counts())
```

```python
# Check CLTV for nulls too (since we never explicitly checked it)
print("Nulls in CLTV:", df['CLTV'].isnull().sum())
```

```python
#  Full missing-value check across ALL columns (final safety net)
print(df.isnull().sum().sum())
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# once agin check for dublicate values
print("Duplicate rows:", df.duplicated().sum())
```

```python
df_encoded = pd.get_dummies(df, drop_first=True)
print(df_encoded.shape)
```

```python
# Final verification before saving (this is the step that would have caught the earlier problem)
print("Total nulls in final dataset:", df_encoded.isnull().sum().sum())
assert df_encoded.isnull().sum().sum() == 0, "Still have nulls — do not save yet!"
print("Clean. Safe to save.")
```

```python
correlations = df_encoded.corr()['Churn'].sort_values(ascending=False)
print(correlations)
```
